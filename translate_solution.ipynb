```python
import requests
from docx import Document
import os
subscription_key = "9TESeILZYQVRtectZg5yX412hQYaYiXlLmEj1wJV2qVrFkajwVrZJQQJ99BKACHYHv6XJ3w3AAAbACOG7Geo"
endpoint = "https://api.cognitive.microsofttranslator.com/"
location = "eastus2"
language_destination = 'pt-br'

def translation_text(text, target_language):
    api_path = '/translate'
    constructed_url = endpoint + api_path
    headers = {
        'Ocp-Apim-Subscription-Key': subscription_key,
        'Ocp-Apim-Subscription-Region': location,
        'X-ClienttraceId': str(os.urandom(16)),
        'Content-type': 'application/json'
    }

    body = [{
        'text' : text
    }]
    params = {
        'api-version': '3.0',
        'from': 'en',
        'to' : target_language
    }
    request = requests.post(constructed_url, headers=headers, json=body, params=params)
    response_json = request.json()
    return response_json[0]["translations"][0]["text"]
```

```python
def translate_document (file_path, target_lang):
  document = Document(file_path)
  full_text = []
  for paragraph in document.paragraphs:
    # Call the global translation_text function and pass the correct target language
    translated_paragraph = translation_text(paragraph.text, target_lang)
    full_text.append(translated_paragraph)

  translated_doc = Document()
  for line in full_text:
    # Use the correct method to add paragraphs
    translated_doc.add_paragraph(line)

  # Construct the output path outside the loop and use target_lang
  path_translated = file_path.replace(".docx", f"_{target_lang}.docx")
  translated_doc.save(path_translated)
  return path_translated

# Define input_file before calling the function
input_file = "/content/musica_there.docx"
# target_language is a global variable defined in a previous cell
# Call the function with correct argument names that match the definition
translated_file = translate_document(file_path=input_file, target_lang=target_language)
print(f"Translated document saved to: {translated_file}")
```
